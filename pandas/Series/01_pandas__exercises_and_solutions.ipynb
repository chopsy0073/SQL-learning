```python
import numpy as np
import pandas as pd
```

## Series Creation

```python
# Create a Series of 5 student names with custom roll numbers as indexes.
name = ['aaa','bbb','ccc','ddd','eee']
roll_no = [101,102,103,104,105]

pd.Series(name,index=roll_no)
```

```python
# Create a Series from a dictionary containing 5 product names and prices.
dict={
    'product1' : 20,
    'product2' : 30,
    'product3' : 40,
    'product4' : 50,
    'product5' : 80
    }
marks_Series = pd.Series(dict)
marks_Series
```

```python
# Create a Series of monthly sales data for one year.
monthly_sales = [100,200,300,400,500,600,700,800,900,1000,1100,1200]
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
pd.Series(monthly_sales,index=months)
```

## Statistical Functions

```python
# Create a Series of 20  stu_marks and find mean, median, max, and min.
stu_marks =[10,20,40,30,44,55,63,22,45,23,45,67,86,43,33,54,65,63,86,74]
print(np.mean(stu_marks))
print(np.median(stu_marks))
print(np.max(stu_marks))
print(np.min(stu_marks))
```

```python
# Find the standard deviation and variance of employee salaries.
salaries = [50000,60000,70000,80000,90000]
print(np.std(salaries))
print(np.var(salaries))
```

## Value Counts

```python
# Create a Series of fruits and find frequency of each fruit
fruits = pd.Series(['Apple','Mango','Apple','Banana','Mango','Apple','Banana'])
print(fruits.value_counts())
```

## Data Modification

```python
# Update a specific value using index label.
s = pd.Series([50,60,70], index=['A','B','C'])
s['B'] = 80
print(s)
```

```python
marks = [50,30,40,60]
```

```python
# Increase all marks by 5
marks_series = pd.Series(marks)
new_marks = marks_series + 5
print(new_marks)
```

```python
# Find students score above 40
marks_series = pd.Series(marks)
print(marks_series[marks_series > 40])
```

## Boolean Indexing

```python
# Find all students scoring above 50.
marks_series = pd.Series(marks)
print(marks_series[marks_series > 50])
```

```python
# Find values between 40 and 80.
marks_series = pd.Series(marks)
print(marks_series[(marks_series > 40) & (marks_series < 80)])
```

## Missing Values

```python
# Create a Series with NaN values, remove them.
series  = pd.Series([1,2,3,np.nan,4,np.nan,5])
print(series.isnull().sum())
print(series.dropna())
```
