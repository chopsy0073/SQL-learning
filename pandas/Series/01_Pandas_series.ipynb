```python
import numpy as np
import pandas as pd
```

## Series from lists

```python
# string
country = ['India','pakistan','USA','Nepal','srilanka']
pd.Series(country)
```

```python
# integers
runs = [13,24,56,23,4,5]
pd.Series(runs)
```

```python
# custom index
marks = [64,34,55,67,100]
subject = ['maths','science','english','hindi','biology']

pd.Series(marks,index=subject)
```

```python
# set a name
marks = pd.Series(marks, index=subject, name='Aditya ke marks')
marks
```

## Series from dict

```python
marks = {
    'maths' :67,
    'english' :57,
    'science' :89,
    'hindi' :100
}
marks_Series = pd.Series(marks,name="Aditya ke marks")
marks_Series
```
